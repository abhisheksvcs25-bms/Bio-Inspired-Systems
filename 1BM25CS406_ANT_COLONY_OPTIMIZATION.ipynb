```python

import random

graph = {
    'A': {'B': 2, 'C': 4},
    'B': {'A': 2, 'C': 1, 'D': 5},
    'C': {'A': 4, 'B': 1, 'D': 2, 'E': 3},
    'D': {'B': 5, 'C': 2, 'E': 1, 'F': 4},
    'E': {'C': 3, 'D': 1, 'F': 2},
    'F': {'D': 4, 'E': 2}
}

source = 'A'
destination = 'F'
ants = 20
iterations = 100
alpha = 1
beta = 5
evaporation = 0.1
Q = 100

pheromone = {
    u: {v: 1.0 for v in graph[u]} for u in graph
}

best_path = None
best_cost = float('inf')

for _ in range(iterations):
    paths = []

    for _ in range(ants):
        path = [source]
        current = source
        cost = 0

        while current != destination:
            choices = [
                node for node in graph[current]
                if node not in path
            ]

            if not choices:
                break

            weights = [
                pheromone[current][node] ** alpha
                * (1 / graph[current][node]) ** beta
                for node in choices
            ]

            next_node = random.choices(
                choices, weights=weights, k=1
            )[0]

            cost += graph[current][next_node]
            path.append(next_node)
            current = next_node

        if current == destination:
            paths.append((path, cost))

            if cost < best_cost:
                best_path = path[:]
                best_cost = cost

    for u in graph:
        for v in graph[u]:
            pheromone[u][v] *= (1 - evaporation)

    for path, cost in paths:
        deposit = Q / cost
        for i in range(len(path) - 1):
            u, v = path[i], path[i + 1]
            pheromone[u][v] += deposit
            pheromone[v][u] += deposit

if best_path:
    print("Best Route:", " -> ".join(best_path))
    print("Total Route Cost:", best_cost)
else:
    print("No route found.")
```
