```python
import random

graph = {
    1: {2: 2, 3: 5, 4: 7},
    2: {1: 2, 4: 3},
    3: {1: 5, 4: 3, 5: 4},
    4: {1: 7, 2: 3, 3: 3, 6: 2},
    5: {3: 4, 6: 2},
    6: {4: 2, 5: 2}
}

source = int(input("Enter source node: "))
destination = int(input("Enter destination node: "))

def find_paths(current, destination, path):
    if current == destination:
        return [path]

    paths = []

    for next_node in graph[current]:
        if next_node not in path:
            paths += find_paths(next_node, destination, path + [next_node])

    return paths

def cost(path):
    total = 0

    for i in range(len(path) - 1):
        total += graph[path[i]][path[i + 1]]

    return total

def fitness(path):
    return 1 / (1 + cost(path))

paths = find_paths(source, destination, [source])

if not paths:
    print("No path exists.")
    exit()

population = random.sample(paths, min(6, len(paths)))

generations = 20

for generation in range(generations):

    population.sort(key=cost)

    new_population = population[:2]

    while len(new_population) < len(population):

        parent1 = random.choice(population)
        parent2 = random.choice(population)

        common = list(set(parent1[1:-1]) & set(parent2[1:-1]))

        if common:
            point = random.choice(common)
            i = parent1.index(point)
            j = parent2.index(point)

            child = parent1[:i] + parent2[j:]

            if child[-1] == destination and len(child) == len(set(child)):
                new_population.append(child)
            else:
                new_population.append(random.choice(population))
        else:
            new_population.append(random.choice(population))

    population = new_population

    if random.random() < 0.3:
        index = random.randint(2, len(population) - 1)
        population[index] = random.choice(paths)

best_path = min(population, key=cost)

print("\nOptimal Path:", " -> ".join(map(str, best_path)))
print("Minimum Cost:", cost(best_path))
```
